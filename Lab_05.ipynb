# **Lab Task 01**

## (a) Task 1.1 — Write tokenize(sentence)

```python
def tokenize(sentence):
    words = sentence.lower().split()
    return ["<s>"] + words + ["</s>"]
```

## (b) Task 1.2 — Tokenize every training and test sentence

```python
train = [
    "the cat sat on the mat", "the cat ate the fish",
    "the dog sat on the rug", "the dog ate the bone"
]
test = ["the cat sat on the rug", "the fish sat on the mat"]

train_tokens = [tokenize(sentence) for sentence in train]
test_tokens = [tokenize(sentence) for sentence in test]

print("Training:")
for sentence, tokens in zip(train, train_tokens):
    print(sentence, "->", tokens)
print("\nTest:")
for sentence, tokens in zip(test, test_tokens):
    print(sentence, "->", tokens)
```

## (c) Task 1.3 — Count training tokens with and without markers

```python
with_markers = sum(len(tokens) for tokens in train_tokens)
without_markers = sum(len(sentence.lower().split()) for sentence in train)
print("With markers:", with_markers)
print("Without markers:", without_markers)
```

## (d) Task 1.4 — Build and print vocabulary V

```python
def build_vocab(sentences):
    vocab = set()
    for sentence in sentences:
        vocab.update(sentence.lower().split())
    vocab.add("</s>")
    return vocab

Vocab = build_vocab(train)
V = len(Vocab)
print("V =", V)
print("Sorted vocabulary:", sorted(Vocab))
```

# **Lab Task 02**

## (a) Task 2.1 — Write get_ngrams(tokens, n)

```python
def get_ngrams(tokens, n):
    if n <= 0:
        raise ValueError("n must be at least 1")
    return [tuple(tokens[i:i+n]) for i in range(len(tokens) - n + 1)]
```

## (b) Task 2.2 — Print unigrams, bigrams and trigrams of first training sentence

```python
first_sentence = train_tokens[0]
print("Tokens:", first_sentence)
for n in (1, 2, 3):
    print(f"{n}-grams:", get_ngrams(first_sentence, n))
```

## (c) Task 2.3 — Verify the N-gram count formula for every sentence

```python
for tokens in train_tokens + test_tokens:
    m = len(tokens)
    for n in (1, 2, 3):
        actual = len(get_ngrams(tokens, n))
        expected = max(0, m - n + 1)
        print(f"m={m}, n={n}: actual={actual}, expected={expected}")
        assert actual == expected
print("All N-gram count checks passed.")
```

## (d) Task 2.4 — Handle n larger than sentence length

```python
print(get_ngrams(["one", "two"], 4))
```

# **Lab Task 03**

## (a) Task 3.1 — Build unigram, bigram and trigram counts

```python
from collections import Counter

unigram_counts = Counter()
bigram_counts = Counter()
trigram_counts = Counter()

for tokens in train_tokens:
    unigram_counts.update(token for token in tokens if token != "<s>")
    bigram_counts.update(get_ngrams(tokens, 2))
    trigram_counts.update(get_ngrams(tokens, 3))
```

## (b) Task 3.2 — Print count tables from highest to lowest count

```python
def print_counts(title, counts):
    print("\n" + title)
    for ngram, count in sorted(counts.items(), key=lambda item: (-item[1], item[0])):
        print(ngram, ":", count)

print_counts("Unigram counts", unigram_counts)
print_counts("Bigram counts", bigram_counts)
print_counts("Trigram counts", trigram_counts)
```

## (c) Task 3.3 — Print five most common bigrams and distinct N-gram counts

```python
print("Five most common bigrams:", bigram_counts.most_common(5))
print("Distinct unigrams:", len(unigram_counts))
print("Distinct bigrams:", len(bigram_counts))
print("Distinct trigrams:", len(trigram_counts))
```

## (d) Task 3.4 — Write count_of and test seen/unseen bigrams

```python
def count_of(ngram):
    counts = {1: unigram_counts, 2: bigram_counts, 3: trigram_counts}
    return counts[len(ngram)].get(tuple(ngram), 0)

print("Seen ('the', 'cat'):", count_of(("the", "cat")))
print("Unseen ('fish', 'dog'):", count_of(("fish", "dog")))
```

# **Lab Task 04**

## (a) Task 4.1 — Write unigram_prob using MLE

```python
N_unigrams = sum(unigram_counts.values())

def unigram_prob(word):
    return unigram_counts.get(word, 0) / N_unigrams if N_unigrams else 0
```

## (b) Task 4.2 — Print unigram probabilities and assert they sum to 1

```python
for word in sorted(Vocab):
    print(f"{word:10s} {unigram_prob(word):.4f}")
print("Sum:", sum(unigram_prob(word) for word in Vocab))
assert abs(sum(unigram_prob(word) for word in Vocab) - 1) < 1e-10
```

## (c) Task 4.3 — Write unigram_sentence_prob and test both sentences

```python
def unigram_sentence_prob(sentence):
    tokens = tokenize(sentence)[1:]
    probability = 1.0
    for word in tokens:
        probability *= unigram_prob(word)
    return probability

for sentence in test:
    print(sentence, "->", f"{unigram_sentence_prob(sentence):.8f}")
```

## (d) Task 4.4 — Show different word orders can have equal unigram probability

```python
from itertools import permutations

words = test[0].lower().split()
permutation_probs = []
for order in permutations(words):
    probability = 1.0
    for word in order:
        probability *= unigram_prob(word)
    probability *= unigram_prob("</s>")
    permutation_probs.append((" ".join(order), probability))

for sentence, probability in permutation_probs[:5]:
    print(sentence, "->", f"{probability:.8f}")
print("Distinct probabilities:", len(set(prob for _, prob in permutation_probs)))
```

# **Lab Task 05**

## (a) Task 5.1 — Write bigram_prob using MLE

```python
bigram_context_counts = Counter()
for (prev, word), count in bigram_counts.items():
    bigram_context_counts[prev] += count

def bigram_prob(prev, word):
    denominator = bigram_context_counts.get(prev, 0)
    return bigram_counts.get((prev, word), 0) / denominator if denominator else 0
```

## (b) Task 5.2 — Write next_word_distribution, print distribution for 'the', and assert sum

```python
def next_word_distribution(prev):
    denominator = bigram_context_counts.get(prev, 0)
    if denominator == 0:
        return {}
    return {
        word: bigram_counts[(prev, word)] / denominator
        for word in sorted(Vocab) if bigram_counts.get((prev, word), 0) > 0
    }

distribution = next_word_distribution("the")
print(distribution)
assert abs(sum(distribution.values()) - 1) < 1e-10
```

## (c) Task 5.3 — Write bigram_sentence_prob and test both sentences

```python
def bigram_sentence_prob(sentence):
    tokens = tokenize(sentence)
    probability = 1.0
    for prev, word in get_ngrams(tokens, 2):
        probability *= bigram_prob(prev, word)
    return probability

for sentence in test:
    print(sentence, "->", f"{bigram_sentence_prob(sentence):.8f}")
```

## (d) Task 5.4 — Write predict_next with alphabetical tie-breaking and test six words

```python
def predict_next(prev):
    distribution = next_word_distribution(prev)
    if not distribution:
        return None
    highest = max(distribution.values())
    return sorted(word for word, prob in distribution.items() if prob == highest)[0]

for word in ["the", "cat", "dog", "fish", "sat", "ate"]:
    print(word, "->", predict_next(word))
```

# **Lab Task 06**

## (a) Task 6.1 — Write trigram_prob

```python
trigram_context_counts = Counter()
for (w1, w2, w3), count in trigram_counts.items():
    trigram_context_counts[(w1, w2)] += count

def trigram_prob(w1, w2, w3):
    denominator = trigram_context_counts.get((w1, w2), 0)
    return trigram_counts.get((w1, w2, w3), 0) / denominator if denominator else 0
```

## (b) Task 6.2 — Write trigram_sentence_prob and test both sentences

```python
def trigram_sentence_prob(sentence):
    tokens = tokenize(sentence)
    if len(tokens) < 3:
        return 1.0
    probability = bigram_prob(tokens[0], tokens[1])
    for i in range(2, len(tokens)):
        probability *= trigram_prob(tokens[i-2], tokens[i-1], tokens[i])
    return probability

for sentence in test:
    print(sentence, "->", f"{trigram_sentence_prob(sentence):.8f}")
```

## (d) Task 6.4 — List unseen bigrams/trigrams and compare their counts

```python
for sentence in test:
    tokens = tokenize(sentence)
    unseen_bi = [ng for ng in get_ngrams(tokens, 2) if bigram_counts.get(ng, 0) == 0]
    unseen_tri = [ng for ng in get_ngrams(tokens, 3) if trigram_counts.get(ng, 0) == 0]
    print("\nSentence:", sentence)
    print("Unseen bigrams:", unseen_bi)
    print("Unseen trigrams:", unseen_tri)
    if len(unseen_bi) > len(unseen_tri):
        print("More unseen: bigrams")
    elif len(unseen_tri) > len(unseen_bi):
        print("More unseen: trigrams")
    else:
        print("Equal number of unseen bigrams and trigrams")
```

```python
def bigram_prob_addone(prev, word):
    return (bigram_counts.get((prev, word), 0) + 1) / (
        bigram_context_counts.get(prev, 0) + V
    )
```

# **Lab Task 09**

## (a) Task 9.1 — Write perplexity; return infinity if any probability is zero

```python
from math import exp, inf

def perplexity(sentences, prob_fn):
    probabilities = [prob_fn(sentence) for sentence in sentences]
    if any(p == 0 for p in probabilities):
        return inf
    token_count = sum(len(tokenize(sentence)) - 1 for sentence in sentences)
    log_probability = sum(log(p) for p in probabilities)
    return exp(-log_probability / token_count)
```

## (b) Task 9.2 — Compare perplexity for bigram MLE, bigram add-one and unigram add-one

```python
def sentence_probability_bigram_addk(sentence, k):
    tokens = tokenize(sentence)
    probability = 1.0
    for prev, word in get_ngrams(tokens, 2):
        probability *= bigram_prob_addk(prev, word, k)
    return probability

def sentence_probability_unigram_addone(sentence):
    tokens = tokenize(sentence)[1:]
    probability = 1.0
    for word in tokens:
        probability *= (unigram_counts.get(word, 0) + 1) / (N_unigrams + V)
    return probability

models = [
    ("Bigram MLE", lambda s: bigram_sentence_prob(s)),
    ("Bigram add-one", lambda s: sentence_probability_bigram_addk(s, 1)),
    ("Unigram add-one", lambda s: sentence_probability_unigram_addone(s))
]
print(f"{'Model':20s} {'Perplexity':>14s}")
for name, fn in models:
    print(f"{name:20s} {perplexity(test, fn):14.4f}")
```

## (c) Task 9.3 — Compute add-k perplexities and choose best k

```python
k_results = []
for k in [1, 0.5, 0.1, 0.01]:
    value = perplexity(test, lambda sentence, k=k: sentence_probability_bigram_addk(sentence, k))
    k_results.append((k, value))

print(f"{'k':>8s} {'Perplexity':>14s}")
for k, value in k_results:
    print(f"{k:8.2f} {value:14.4f}")

best_k, best_test_perplexity = min(k_results, key=lambda row: row[1])
print("Best k on test set:", best_k)
print("Best test perplexity:", f"{best_test_perplexity:.4f}")
```

## (d) Task 9.4 — Select k on validation data, then report test perplexity

```python
validation_train = train[:3]
validation_sentence = train[3]
validation_tokens = [tokenize(sentence) for sentence in validation_train]

validation_bigram_counts = Counter()
for tokens in validation_tokens:
    validation_bigram_counts.update(get_ngrams(tokens, 2))

validation_context_counts = Counter()
for (prev, word), count in validation_bigram_counts.items():
    validation_context_counts[prev] += count

def validation_bigram_addk(prev, word, k):
    return (validation_bigram_counts.get((prev, word), 0) + k) / (
        validation_context_counts.get(prev, 0) + k * V
    )

def validation_sentence_probability(sentence, k):
    tokens = tokenize(sentence)
    probability = 1.0
    for prev, word in get_ngrams(tokens, 2):
        probability *= validation_bigram_addk(prev, word, k)
    return probability

validation_results = []
for k in [1, 0.5, 0.1, 0.01]:
    value = perplexity([validation_sentence],
                       lambda sentence, k=k: validation_sentence_probability(sentence, k))
    validation_results.append((k, value))

print("Validation sentence:", validation_sentence)
print(f"{'k':>8s} {'Validation perplexity':>24s}")
for k, value in validation_results:
    print(f"{k:8.2f} {value:24.4f}")

best_validation_k, best_validation_perplexity = min(validation_results, key=lambda row: row[1])
validation_selected_test_perplexity = perplexity(
    test, lambda sentence: sentence_probability_bigram_addk(sentence, best_validation_k)
)

print("Best k on validation:", best_validation_k)
print("Validation perplexity:", f"{best_validation_perplexity:.4f}")
print("Test perplexity using validation-selected k:", f"{validation_selected_test_perplexity:.4f}")
print("Best k selected directly on test:", best_k)
print("Test perplexity using test-selected k:", f"{best_test_perplexity:.4f}")
print("Selecting k on validation data is preferable because the test set remains for final evaluation.")
```
