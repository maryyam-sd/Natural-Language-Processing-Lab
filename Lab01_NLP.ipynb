```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
for t in tokens:
    if t.isalpha():
        label = "alphabetic"
    elif t.isdigit():
        label = "numeric"
    else:
        label = "mixed"
    print(f"{t}: {label}")
```

### **(c).**

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words, pure_numbers, mixed_tokens = [], [], []
for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print("pure_words:", pure_words)
print("pure_numbers:", pure_numbers)
print("mixed_tokens:", mixed_tokens)
```

### **(d).**

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    return "MIXED"

for t in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(f"{t}: {validate_token(t)}")
```

## **Task 08**

### **(a).**

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = "  HELLO Natural Language Processing  "
print(clean(sample))
print(clean_lambda(sample))
```

### **(b).**

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
toks = tokenize(sentence.lower())
print(remove_short(toks, 3))
```

### **(c).**

```python
def word_lengths(tokens):
    return {word: len(word) for word in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print(word_lengths(tokens))
```

### **(d).**

```python
def nlp_pipeline(text, min_len):
    cleaned = text.strip().lower()
    toks = cleaned.split()
    filtered = [t for t in toks if len(t) > min_len]
    return {t: len(t) for t in filtered}

print(nlp_pipeline("I love working with NLP tools in Python daily", 3))
print(nlp_pipeline("Deep learning models process text efficiently", 4))
```

### **(a).**

```python
from collections import Counter

words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lower_words = list(map(lambda w: w.lower(), words))
print(lower_words)

lengths = list(map(lambda w: len(w), words))
print(lengths)
```

### **(b).**

```python
words_long = list(filter(lambda w: len(w) > 3, words))
print("words_long:", words_long)

words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print("words_with_t:", words_with_t)
```

### **(c).**

```python
print("Alphabetical:", sorted(words))
print("By length ascending:", sorted(words, key=len))
print("By length descending:", sorted(words, key=len, reverse=True))
```
